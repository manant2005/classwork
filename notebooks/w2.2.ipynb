```python
# Integers

a: int = 26
b: int = 5
print(f"addition: {a + b}") #addition
print(f"division: {a/b}") #division
print(f"multiply {a*b}") #multiplication
print(f"subtraction {a-b}") #subtraction
print(f"exponentiation {a**b}") #exponentiation
print(f"modulus {a % b}") #modulus
print (f"floor division {a // b}") #floor division
```

```python
a = str (a)
b = float(b)

print(f"new type a and b: {a.zfill(7), b}")
print(int(3.99999))
```

```python
# booleans
x: bool = True
print(f"x = {x}")
print(f"b as bool: {bool(b)}")
print(f"0 as bool: {bool(0)}")
print(f"empty string as bool {bool("")}")
print(f'a string as bool: {bool(a)}')

```

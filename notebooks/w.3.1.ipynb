```python
# Chess with Lists

Pawns = ["P"] * 8
Cool = ["R", "N", "B", "Q", "K", "B", "N", "R"]
empty = [" "] * 8

chess = [
    Cool.copy(),
    Pawns.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    Pawns.copy(),
    Cool.copy(),
    
]
chess
```

```python
chess[0][1] = " "
chess[2][2] = "N"
chess
```

```python
# tuple

# not mutable
# allows duplicates
# indexed 
tpl = (1,2,2,3, "something")
tpl2 = tuple((1,2,3,4,5))
print(tpl)
print(tpl2)
```

```python
tpl.count (2)
tpl.index(2)
```

```python
# set
# can change (mutable)
# no duplicates
# unordered

set1 = {1,2,1,2,1,2,3} # the bad way
set2 = set([1,2,3,3,3,4,4,4,4])
set2
```

```python
set2.remove(3)
len(set2)
```

```python
set2
```

```python
set2.difference(set1)
```

```python
set2 - set1
```

```python
# dictionary
# key: value pairs
# indexed by its key (not by its position)
# can change (mutable)
# keys cannot duplicate, values can 
edwin = {
    "name": "edwin",
    "city" : "charlotte",
    "age" : 8
    
}


person = {
    "name": "Nate",
    "age" : 40,
    "city" : "charlotte",
    "children": ["edwin", "preston"]}
    
preston = {
    "name": "preston",
    "age" : 12,
    "city" : "charlotte",
}
person
edwin
```
